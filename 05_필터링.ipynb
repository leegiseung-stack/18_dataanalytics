```python
import pandas as pd
import seaborn as sns

titanic=sns.load_dataset("titanic")
df=titanic.loc[0:9,["age","fare"]]    #loc 행이름과 열이름 ,기준으로 행,
```

```python
#불린 인덱싱
df["age"] <20  #비교연산자 결과값은 트루 펄스 
```

```python
#필터링 (아주중요!)
cond= df["age"] <20 
df[cond]
df.loc[cond]
#둘다가능

```

```python
cond= df["age"] <20 
df[~cond] #물결표시일시 False인거 보여줌 
```

```python
#나이가 10대인 승객
#cond=df[10<=age<20]
(df["age"]>=10) & (df["age"]<20)
```

```python
cond=(df["age"]>=10) & (df["age"]<20)
df[cond]
```

```python
#titanic.age
cond= (titanic["age"] >= 10) & (titanic.age <20)
titanic[cond]
```

```python
#나이가 10세미만이고 여성인 승객
#titanic.head()
(titanic["age"] > 10) & (titanic["sex"] == "female")

titanic[(titanic["age"] > 10) & (titanic["sex"] == "female")]
```

```python
#나이가 10세 미만 이고 또는 60세 이상인 승객 
cond=(titanic["age"] < 10) | (titanic["age"]>=60) #합집합 두개는 동싱에충족x
titanic[cond]
```

```python
titanic["fare"][cond]
```

```python
df_under10 = titanic.loc[cond,["age","sex","alone"]]
df_under10
```

```python
titanic[["age","sex","alone"]][cond]
```
