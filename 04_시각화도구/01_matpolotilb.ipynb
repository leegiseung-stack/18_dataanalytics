```python
import matplotlib.pyplot as plt

# 스타일 서식 지정
plt.style.use("ggplot")

# 그림 사이즈 늘리기 (figaize -> figsize)
plt.figure(figsize=(14, 5))

# x축 눈금 회전하기 (pit -> plt)
plt.xticks(size=10, rotation="vertical")

# X축, y축 데이터를 plot함수에 입력 (polt -> plot)
plt.plot(sr_one.index, sr_one.values, marker='o', markersize=10)
plt.title("서울->경기 인구 이동", size=30)

plt.xlabel("기간", size=20)
plt.ylabel("이동인구수", size=20)

# 범례 표시 (pit -> plt)
plt.legend(labels=["서울->경기"], loc='best', fontsize=15)

# y축 범위 지정 (tlim -> ylim)
plt.ylim(50000, 800000)

# 주석 표시 - 화살표 1 (xy= 형태 확인)
plt.annotate('', xy=(20, 620000),
             xytext=(2, 290000),
             xycoords="data",
             arrowprops=dict(arrowstyle='->', color="skyblue", lw=5)) # 화살표 서식

# 주석 표시 - 화살표 2 (xy=(...) 형태로 수정)
plt.annotate('', xy=(47, 450000), # 화살표머리부분(끝점) (xy= 추가)
            xytext=(30, 580000), # 화살표꼬리부분(시작점)
            xycoords="data",
            arrowprops=dict(arrowstyle='->', color="olive", lw=5)) # 좌표체계

# 주석 텍스트 1 (xy=(...) 형태로 수정)
plt.annotate("인구이동 증가(1970-1995)", # 연도 오타(1955->1995)도 같이 수정하면 좋습니다
            xy=(10, 350000), rotation=25, # xy= 추가
            va="baseline", ha="center", fontsize=15)

# 주석 텍스트 2 (xy=(...) 형태로 수정)
plt.annotate("인구이동 감소(1995-2017)",
            xy=(40, 470000), rotation=11, va="baseline", ha="center", fontsize=15) # xy= 추가

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.style.use("ggplot")
fig = plt.figure(figsize=(20, 5))
ax = fig.add_subplot(1, 1, 1)  # 차트 생성

# 1. 오타 수정 (maker -> marker, lable -> label)
ax.plot(
    sr_one,
    marker="o",
    markerfacecolor="orange",
    markersize=10,
    color="olive",
    linewidth=2,
    label="서울->경기",
)

ax.legend(loc="best")

# 2. 오타 수정 (set_yilm -> set_ylim)
ax.set_ylim(50000, 800000)

ax.set_title("서울->경기 인구 이동", size=20)
ax.set_xlabel("기간", size=12)
ax.set_ylabel("이동 인구수", size=12)
ax.set_xticks(sr_one.index)
ax.set_xticklabels(sr_one.index,rotation=75)
ax.tick_params(axis="x",labelsize=10)
ax.tick_params(axis="y",labelsize=10)
# 3. x축 눈금 설정 수정 (필요한 경우 아래와 같이 사용)
# 만약 index를 x축 라벨로 쓰고 싶다면:
# ax.set_xticklabels(index, rotation=90)  # 글씨가 겹치면 회전 추가 가능
```

```python
col_years= list(map(str,range(1970,2018)))
df_3=df_seoul.loc[["충청남도","경상북도","강원도"],col_years]
plt.style.use("ggplot")
#그래객체생성(gigure에 1개의 서브 플롯을 생성)
fig=plt.figure(figsize=(20,5))
ax=fig.add_subplot(1,1,1)
#axe객체에 plot함수로 그래프 출력
ax.plot(col_years,df_3.loc["충청남도",:],marker="o",markerfacecolor="green",markersize=10,color="olive",
        linewidth=2,label="서울->충남")
ax.plot(col_years,df_3.loc["경상북도",:],marker="o",markerfacecolor="skyblue",markersize=10,color="olive",
        linewidth=2,label="서울->경북")
ax.plot(col_years,df_3.loc["강원도",:],marker="o",markerfacecolor="red",markersize=10,color="olive",
        linewidth=2,label="서울->강원")

ax.legend(loc="best")
#차트제목추가
ax.set_title("서울->충남,경북,강원인구이동",size=20)
#축이름추가
ax.set_xlabel("기간",size=12)
ax.set_ylabel("이동인구수",size=12)

#축 눈금의 위치 지정
ax.set_xticks(range(len(col_years)))
#축 눈금 레이블 지정 및 90도 회전
ax.set_xticklabels(col_years,rotation=90)
#축눈금 레이블 크기
ax.tick_params(axis="x",labelsize=10)
ax.tick_params(axis="y",labelsize=10)
```

```python
#
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import font_manager, rc 
font_path = "./data/malgun.ttf"
font_name = font_manager.FontProperties(fname=font_path).get_name()
rc('font', family=font_name)

plt.style.use("ggplot")
plt.rcParams["axes.unicode_minus"]=False

df= pd.read_excel("./data/남북한발전전력량.xlsx")
df=df[5:10]
df
```

```python
df.drop("전력량 (억㎾h)",axis="columns",inplace=True) #인플레이스하면 재할당x 
df.set_index("발전 전력별",inplace=True)
```

```python
df=df.T
df.head()
```

```python
#for col df.columns
for col in df.columns:
    df[col]= df[col].replace("-","0")
print(df.head())
df.dtypes
```

```python
df=df.astype(float)
df.info()
```

```python
df.head()
df= df.rename(columns={"합계":"총발전량"})
df.head()
```

```python
df["총발전량"].shift(1)
```

```python
df["총발전량-1년"]=df["총발전량"].shift(1)
df["증감율"]=((df["총발전량"] / df["총발전량-1년"]) -1) *100
df.head()
```

```python

ax1=df[["수력","화력"]].plot(
    kind="bar",
    figsize=(20,10),
    width=0.7,
    stacked=True #같이쌓는거 False면 다른거 막대두개나타남
)

ax2= ax1.twinx()
ax2.plot(
    df.index,
    df["증감율"],
    ls="--",
    marker="o",
    markersize=20,
    color="green",
    label="정년대비증감율"
)

ax1.set_ylim(0,500)
ax2.set_ylim(-50,30)
plt.show()
```

```python
import pandas as pd 
import matplotlib.pyplot as plt

df=pd.read_csv("./data/auto-mpg.csv",header=None)
df.columns = ['mpg', 'cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model year', 'origin', 'car name']
df["mpg"].plot(
    kind="hist",
    bins=10,
    color="coral",
    figsize=(10,5)

)

plt.show
```

```python
df.head()
```

```python
df[["mpg","origin"]].plot(
    by=["origin"],
    kind="hist",
    figsize=(8,10)
)
```

```python
df[["mpg","origin","cylinders"]].plot(
    by=['origin','cylinders'],kind="hist",figsize=(8,30)
)
```

```python
#산점도
plt.style.use("default")
df.plot(
    kind="scatter",
    x="weight",
    y="mpg",
    c="coral",
    s=10,
    figsize=(10,5)
)
```

```python
#산점도
plt.style.use("default")
cylinders_size=(df.cylinders / df.cylinders.max())*300

df.plot(
    kind="scatter",
    x="weight",
    y="mpg",
    c="coral",
    s=cylinders_size,
    figsize=(10,5),
    alpha=0.3

)
```

```python
#산점도
plt.style.use("default") #눈금표시없애기
cylinders_size=(df.cylinders / df.cylinders.max())*300

df.plot(
    kind="scatter",
    x="weight",
    y="mpg",
    marker="+",
    s=50,
    figsize=(10,5),
    alpha=0.3,
    cmap="viridis",
    c=cylinders_size
    
)
```
