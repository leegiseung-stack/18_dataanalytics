```python
#선그래프
import pandas as pd 
df=pd.read_excel("./data/남북한발전전력량.xlsx")
df.head()

df_ns=df.iloc[[0,5],3:]
df_ns.index=["south","north"]
df_ns
df_ns.info()
```

```python
df_ns=df_ns.astype(int)
df_ns.info()
#결손값이없어서 int형으로 변환됨
df_ns
```

```python
#df_ns.plot() #그래프를그려주는매서드
df_ns.plot()
```

```python
tdf_ns= df_ns.transpose() #행과 열을 바꿈 
tdf_ns.plot()
```

```python
#막대그래프
tdf_ns.plot(kind="bar")
```

```python
tdf_ns.plot(kind="barh") #가로막대
```

```python
tdf_ns.plot(kind="bar",stacked=True) #누적세로막대
```

```python
#히스토그램
tdf_ns.plot(kind="hist")
```

```python
tdf_ns.plot(
    kind="hist",
    alpha=0.5, #투명도
    figsize=(6,2), #그림판크기
    grid=True, #격자무늬
    color=["coral","skyblue"],
    edgecolor="black"
)
```

```python
tdf_ns.plot(
    kind="hist",
    alpha=0.8, #투명도
    figsize=(6,2), #그림판크기
    grid=True, #격자무늬
    color=["coral","skyblue"],
    edgecolor="black"
)
```

```python
tdf_ns.plot(
    kind="hist",
    alpha=0.8, #투명도
    figsize=(6,2), #그림판크기
    grid=False, #격자무늬
    color=["blue","coral"],
    edgecolor="black"
)
```

```python
tdf_ns.plot(
    kind="hist",
    alpha=0.8, #투명도
    figsize=(6,2), #그림판크기
    grid=False, #격자무늬
    color=["blue","coral"],
    edgecolor="red"
)
```

```python
tdf_ns.hist(figsize=(6,2)) #하나의캔버스
```
