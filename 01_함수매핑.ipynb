```python
#시리즈원소에 함수 매핑
#series객체.apply(매핑함수)

import seaborn as sns
import pandas as pd

titanic=sns.load_dataset("titanic")
df=titanic.loc[:, ["age","fare"]]
df.head()
df=titanic.loc[0:8, :]
df

```

```python
def add_10(n):
    return n+10

def add_two_abj(a,b):
    return a+b
```

```python
#시리즈객체 함수매핑
#데이터프레임 각열에 함수매핑
def calculae_state(col):
    max_val= col.max()
    min_val = col.min()
    mean_val= col.mean()
    median_val = col.median()
    return pd.Series(
        [max_val,min_val,mean_val,median_val],
        index=["Max","Min","Mean","Median"]
    )

result_df=df.apply(calculae_state,axis=0)
result_df
```

```python
result_sr=df.apply(lambda x:x.max()-x.min())
result_sr
```

```python
def calculate_diff_avg(row):
    diff = row.max() - row.min()
    avg = row.mean()

    return pd.Series([diff,avg],index=["차이","평균"])

result_df2=df.apply(calculate_diff_avg,axis=1)
result_df2
```

```python
titanic.head()
df1=titanic.loc[:,["age","fare"]] #콤마를기준으로 행 :는 모든 행을다가져오겠다
df2=titanic.loc[:,["embark_town","embarked"]] #콤마를기준으로 행 :는 모든 행을다가져오겠다
def missing_value(df):
    return df.isnull()

result_df=df1.pipe(missing_value)
result_df
```

```python
def missing_count(df):
    return missing_value(df).sum()
result_series=df1.pipe(missing_count)
result_series
```
