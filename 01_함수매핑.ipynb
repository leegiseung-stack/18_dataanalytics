```python
#시리즈원소에 함수 매핑
#series객체.apply(매핑함수)

import seaborn as sns
import pandas as pd

titanic=sns.load_dataset("titanic")
df=titanic.loc[:, ["age","fare"]]
df.head()

```

```python
def add_10(n):
    return n+10

def add_two_abj(a,b):
    return a+b
```

```python
print(add_10(10))
print(add_two_abj(10,10))
```

```python
df["age"].head()
```

```python
sr1=df["age"].apply(add_10)
#함수안에 함수를 호출할때는 바로실행하는게아니고 마지막에 실행되길원하기때문에 add_10 만해줌 add_10() X
sr1.head()
```

```python
#람다함수 
sr2=df["age"].apply(lambda n : n+10)
sr2.head()
```

```python
imsi = lambda n : n+10
imsi(df["age"])
```

```python
sr4= df["age"].apply(add_two_abj,b=10)
sr4.head()
```

```python
sr5=df["age"].apply(lambda a,b : a+b ,b=10)
sr5
```
