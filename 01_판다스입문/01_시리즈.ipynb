```python
import pandas as pd
dict_data = {
    "a":1,
    "b":2,
    "c":3
}
sr = pd.Series(dict_data) #시리즈객체
print(type(sr))
print(sr)
```

```python
import pandas as pd
list_data = [
    "2019-01-02" ,3.14 , "ABC" , 100 , True
]

sr = pd.Series(list_data)
print(sr)
```

```python
#시리즈 원소 선택
import pandas as pd

# 중괄호 대신 소괄호() 또는 대괄호[]를 사용하세요
tup_data = ("영인", "2010-05-01", "여", True)  
sr = pd.Series(tup_data, index=["이름", "생년월일", "성별", "학생여부"])

print(sr)
```

```python
#시리즈 원소형태
print(sr["이름"])
print(sr["성별"])
```

```python
#print(sr[[1,2]]) =>더이상 지원안됨
print(sr[["생년월일","성별"]])
```
