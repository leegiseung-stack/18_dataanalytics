```python
from bs4 import BeautifulSoup
import requests
import re
import pandas as pd

url = "https://en.wikipedia.org/wiki/List_of_American_exchange-traded_funds"
resp = requests.get(url, headers={
    "User-Agent" : "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/154.0.0.0 Safari/537.36"
})

soup = BeautifulSoup(resp.text, "lxml")
rows = soup.select("div  ul  li")

etfs = {} # 데이터베이스 저장용 초기화

for row in rows:
    try: 
        # 1. 괄호와 이스케이프(`\`)가 정확히 닫히도록 수정된 정규식
        etf_name = re.findall(r"(.+?)\s\(", row.text)
        etf_market = re.findall(r"\((.+?):", row.text)         
        etf_ticker = re.findall(r":\s(.+?)\)", row.text)
        
        # 2. 리스트 길이를 각각 체크하고 and로 연결
        if len(etf_ticker) > 0 and len(etf_market) > 0 and len(etf_name) > 0:
            print(etf_name)
            etfs[etf_ticker[0]] = [etf_market[0], etf_name[0]]
            
    except AttributeError as err:
        print(err)
        pass

print(etfs)
```
