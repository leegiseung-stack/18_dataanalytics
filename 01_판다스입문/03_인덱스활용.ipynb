```python
#완정중요!
#행인덱스를 기준으로 데이터 프레임 정렬
import pandas as pd

dict_data={
    "c0":[1,2,3],
    "c1":[4,5,6],
    "c2":[7,8,9],
    "c3":[10,11,12],
    "c4":[13,14,15]
}

df=pd.DataFrame(dict_data,index=["r0","r2","r1"])
#DataFreame.객체.sort_index()
df=df.sort_index()
ndf=df.sort_index(ascending=False) #내림차순
#df.sort_index(asscending=True) #오름차순
ndf
```

```python
#열기준 정렬 #자주나옴
#DataFrame객체.sort_values()
import pandas as pd

dict_data={
    "c0":[1,2,3],
    "c1":[4,5,6],
    "c2":[7,8,9],
    "c3":[10,11,12],
    "c4":[13,14,15]
}

df=pd.DataFrame(dict_data,index=["r0","r2","r1"])
df.sort_values("c1",ascending=False)#필드명필요
df.sort_values(by="c2",ascending=True)
df.iloc[0,3]=11
df

df=df.sort_values(by=["c3","c4"],ascending=[False,True])
df
```
