```python
#데이터정규화
import pandas as pd
df=pd.read_csv("./data/auto-mpg.csv",header=None)
df.columns = ['mpg', 'cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model year', 'origin', 'car name']
df.head(3)
mpg_to_kpl=1.60934/3.78941
#킬로미터퍼리터
df["kpl"]=df["mpg"]*mpg_to_kpl
df["kpl"]
```

```python
df["kpl"]=df["kpl"].round(2) 
#소숫점 두번째자리까지 
```

```python
#자료형 변환
df.dtypes
df["horsepower"].unique()
#데이터가 중복되지 않는 값들만 나옴 
```

```python
#?값을 채워줄것임! 
import numpy as np
df["horsepower"]=df["horsepower"].replace("?",np.nan)
#시리즈니까 horsepower로 재할당 
df["horsepower"].unique()
df=df.dropna(subset=["horsepower"],axis=0)
#결측값있는 행지우기 
#데이터프레임이니까 그냥 df로선언해주기 
df["horsepower"].unique()
```

```python
df["horsepower"]=df["horsepower"].astype("float")
#시리즈니까 df["horsepower"]에 재할당
df.dtypes
```

```python
#df["origin"].unique()
df["origin"].value_counts()
```

```python
df["origin"]=df["origin"].replace(
    {
        1:"USA",
        2:"EU",
        3:"JAPAN"
    }
)
#시리즈니까 배열로 재할당 
#df["origin"].value_counts()
#df["origin"].isnull().sum()
df["origin"].astype("category")
```
