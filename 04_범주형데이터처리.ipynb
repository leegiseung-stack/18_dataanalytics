```python
import pandas as pd
import numpy as np
df=pd.read_csv("./data/auto-mpg.csv",header=None)
df.columns = ['mpg', 'cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model year', 'origin', 'car name']
df.head(3)
```

```python
df["horsepower"]=df["horsepower"].replace("?",np.nan)
df=df.dropna(subset=["horsepower"])
df["horsepower"]=df["horsepower"].astype("float")
df.info
```

```python
df["horsepower"].describe()
count,bin_dividers=np.histogram(df["horsepower"],bins=3)
print(count)
print(bin_dividers)
```

```python
bin_names=["저출력","보통출력","고출력"]
df["hp_bin"]=pd.cut(
    x=df["horsepower"],
    bins=bin_dividers,
    labels=bin_names,
    include_lowest=True
)
df["hp_bin"]
```

```python
#더미변수 #원확인코딩(?) ->True /False 해주는것
df["hp_bin"].value_counts()
pd.get_dummies(df["hp_bin"])
```

```python
pd.get_dummies(df["hp_bin"],dtype=float)
```

```python
pd.get_dummies(df["hp_bin"],dtype=float,drop_first=True)
#아니어도 알수있는 자료는 하나삭제해도됨
#ex)남여중 하나만해도 알수있음
#다중공성선회피..(?)

```

```python
from sklearn import preprocessing
from sklearn.preprocessing import LabelEncoder,OneHotEncoder

label_encoder=LabelEncoder()
#매핑시켜주는것 

onehot_encoder = OneHotEncoder()
onehot_labeled= label_encoder.fit_transform(df["hp_bin"].head(15))
#fit_transform 학습시킨데이터 
print(type(onehot_labeled))
print(onehot_labeled)
print(onehot_labeled.ndim)
```

```python
onegot_reshaped=onehot_labeled.reshape(len(onehot_labeled),1)
onehot_fifted=onehot_encoder.fit_transform(onegot_reshaped)
print(onehot_fifted)
```
