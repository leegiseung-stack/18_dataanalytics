```python
# 행삭제:DataFrame객체.drop(행인덱스 또는 배열,axis=0또는 axis="index") #기본은 0
# 열삭제:DataFrame객체.drop(열이름 똔느 배열 , axis=1 또는 axis="colums") 
import pandas as pd

exam_data = {
    "수학":[90,80,70],
    "영어":[98,89,95],
    "음악":[85,95,100],
    "체육":[100,90,90]
}
pd.DataFrame(exam_data,index=["서준","우현","인한"])
```

```python
import pandas as pd

exam_data = {
    "수학":[90,80,70],
    "영어":[98,89,95],
    "음악":[85,95,100],
    "체육":[100,90,90]
}
df=pd.DataFrame(exam_data,index=["서준","우현","인한"])

df2 = df.copy()
df2= df2.drop("우현",axis=0) #axis=0 기본값이라 안써도 자동으로 우현지워짐
df2
```

```python
import pandas as pd

exam_data = {
    "수학":[90,80,70],
    "영어":[98,89,95],
    "음악":[85,95,100],
    "체육":[100,90,90]
}
df=pd.DataFrame(exam_data,index=["서준","우현","인한"])

df3 = df.copy()
df3 = df3.drop(["우현","인한"],axis=0)
df3
```
