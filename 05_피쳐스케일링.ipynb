```python
#크고작은것들을 고르게 만들어줌
import pandas as pd
import numpy as np
df=pd.read_csv("./data/auto-mpg.csv",header=None)
df.columns = ['mpg', 'cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model year', 'origin', 'car name']
df["horsepower"]=df["horsepower"].replace("?",np.nan)
df=df.dropna(subset=["horsepower"])
df["horsepower"]=df["horsepower"].astype("float")
df.info
```

```python
df["horsepower"].describe()
```

```python
#min-max-scaling
df["horse_minmax"]=(df["horsepower"] - df["horsepower"].min())/(df["horsepower"].max()-df["horsepower"].min())
#각각의 홀스파워요소를 최소값을 빼줌
#df["horsepower"] -46 / (230 - 46)
df["horse_minmax"].describe()
```

```python
from sklearn.preprocessing import MinMaxScaler 
#빅분기2번문제
scaler=MinMaxScaler()
df["horse_minmax"]=scaler.fit_transform(df[["horsepower"]])
#민맥스스케일러에 들어가는거는 시리즈가아니라 데이터프레임이라서 대괄호두개를 넣어주세요! 
df["horse_minmax"]
df["horse_minmax"].describe()
```
